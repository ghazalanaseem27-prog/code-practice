# sequence sum

### 1/1! + 2/2! + 3/3! +....

```python
# code

num = int(input("enter a number: "))

result = 0
fact = 1

for i in range(1, num+1):
    fact = fact * i
    result = result + (i/fact)
print(result)
```

# Nested loop

```python
for i in range(0,4):
    for j in range(0,4):
        print(i,j)
```

```python
# (IMP)  patern print

rows=int(input("enter number of rows: "))
for i in range(1,rows+1):
  for j in range(1,i+1):
    print('*',end="")
  print()
```

```python
# pattern 2
rows = int(input("Enter the number of rows: "))
for i in range(1,rows+1):
  for j in range(1,i+1):
    print(j,end="")
  for k in range(i-1,0,-1):
    print(k,end="")

  print()
```

# Loop control statement

### Break
### Continue
### Pass

```python
# Break

for i in range(10):
    if i==5:
        break
    print(i)
```

```python
for i in range(10):
    if i==8:
        break
    print(i)
```

```python
lower=int(input("enter lower range: "))
upper=int(input("enter upper range: "))

for i in range(lower,upper+1):
    for j in range(2,i):
        if i%j==0:
            break
    else:
        print(i)
```

```python
# continue

for i in range(10):
    if i==5:
        continue
    print(i)
```

```python
# pass

for i in range(1,10):
  pass
```

# Accessing subrings from a strings

```python
# Indexing

s='hello world'
print(s[4])
print(s[0])

print(s[-4])
```

```python
# Slicing

a='Hello Ghazala'
print(a[0:6])

print(a[-2:])
print(a[4:0:-1])

print(a[::-1])
```

# Editing and Deleting strings

```python
# editing string

s='hi'
s[0]='H'
# this is not posible because (string are immutable)
```

```python
# deleting string

s='hi'
del s
print(s)
# this is also not possible
```

* Operations on Strings
    * Arithmetic Operations
    * Relational Operations
    * Logical Operations
    * Loops on Strings
    * Membership Operations

```python
# arithmetic operation

print('delhi' + 'mumbai')
print('delhi'*5)
print('*'*50)
```

```python
# relational operator

print('delhi'=='mumbai')
print('delhi'!='mumbai')
print('delhi'>'mumbai')
print('delhi'<'mumbai')
```

```python
# logical operator

print('hello' and 'world')
print('hello' or 'world')
print(not 'hello')
```

```python
# loops on strings

s='hello world'
for i in s:
  print(i)
```

```python
# membership operation

print('D' in 'Delhi')
print('M' in 'Mumbai')
print('D' not in 'Delhi')
print('M' not in 'Mumbai')
```
