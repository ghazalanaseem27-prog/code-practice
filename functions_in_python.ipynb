# Type of arguments
 ###    . Default argument
   ###       .   positional argument
   ###       .   Keyword argument


```python
# default argument

def power(a=1,b=1):
  return a**b
power(2,5)

# if we forget given the value to arguments then we use default function
#this protect the function from error
```

```python
# positionl argument

def add(a,b,c=0):
  return a+b+c
add(1,2,3)
```

```python
# keyword argument

def power(a,b):
  return a**b
power(b=3,a=2)
```

# *args and **kwargs

```python
# *args
# allow us to pass the variable num of keyword argument to a function

def multiply(*args):
  product=1
  for i in args:
    product=product*i
  print(args)
  return product
multiply(1,2,3,4,5,6,7)
```

```python
# **kwargs

def display(**kwargs):
  for (key,value) in kwargs.items():
    print(key,'->',value)
display(india='delhi',srilanka='colombo',nepal='kathmandu')

# **kwargs allow us to pass any  of num keyword argument
# keword argument mean that they contain key-value pairs,like a python dictionary.
```

# Higher order function

```python
# example

def square(x):
  return x**2
#HOF

def transform(f,L):
  output=[]
  for i in L:
    output.append(f(i))
  print(output)
l=[1,2,3,4,5,6,7,8,9]
transform(lambda x:x**2,l)
```

# Map

```python
# square the item of ist

list(map(lambda x:x**2,[1,2,3,4,5]))
```

```python
# odd/even labelling of item list

L = [1,2,3,4,5]
list(map(lambda x:'even'if x%2==0 else'odd',L))
```

```python
# fech names from a list dict

users =[
     {
        'name':'ghazala',
        'age':19,
        'gender':'female'
     },
     {
        'name' : 'ghazal',
        'age' : 20,
        'gender' : 'female'
     },
     {
        'name' : 'gina',
         'age' : 21,
         'gender' : 'female'
     }
]
list(map(lambda p:p['name'],users))
```

# Filter

```python
# num greater then 5
l=[3,4,5,6,7,8,9]
list(filter(lambda x:x>5,l))
```

```python
# fech fruits starting with 'a'
L=['apple','banana','cherry']
list(filter(lambda x:x[0]=='a',L))

fruits=['apple','guva', 'mango']
list(filter(lambda x:x.startswith('a'),fruits))
```
