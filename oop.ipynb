# Write oop classes to handle the following snerious
### . A user can create and view 2D cordinates
### . A user can find out the the distance between two cordinates
### . A user can check if a point lies on a given line
### . A user can find the distance between 2D point and given line

```python
class point:

  def __init__(self,a,b):

     self.a_cor=a
     self.b_cor=b

  def __str__(self):
    return '<{},{}>'.format(self.a_cor,self.b_cor)

  def euclidean__distance(self,other):
    return ((self.a_cor-other.a_cor)**2+(self.b_cor-other.b_cor)**2)**0.5

  def distance_from_origin(self):
    return (self.a_cor**2+self.b_cor**2)**0.5

class line:

  def __init__(self,a,b,c):

    self.a=a
    self.b=b
    self.c=c
  def __str__(self):
    return '{}x + {}y + {} = 0'.format(self.a,self.b,self.c)


  def line__on__point(line,point):
    if line.a*point.a_cor+line.b*point.b_cor+line.c==0:
      return ('point lies on the line')
    else:
      return ('point does not lie on the line')

  def shortes__distance(line,point):
    return abs(line.a*point.a_cor+line.b*point.b_cor+line.c)/(line.a**2+line.b**2)*80.5

p1=point(3,6)
p2=point(4,5)

l1 = line(1,2,3)
p1 = point(1,2)

print(p1)
print(p1.euclidean__distance(p2))
print(p1.distance_from_origin())

print(l1)
print(p1)
print(l1.line__on__point(p1))
print(l1.shortes__distance(p1))
```

```python
class line:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def __str__(self):
        return '{}x + {}y + {} = 0'.format(self.a, self.b, self.c)

    # Naya function do lines ka intersection nikalne ke liye
    def intersection(self, other):
        # Denominator check (a1*b2 - a2*b1)
        denominator = (self.a * other.b) - (other.a * self.b)

        if denominator == 0:
            return "Lines are parallel, they do not intersect!"

        x = ((self.b * other.c) - (other.b * self.c)) / denominator
        y = ((self.c * other.a) - (other.c * self.a)) / denominator

        return point(x, y)

l1 = line(1, -1, 0)  # x - y = 0
l2 = line(1, 1, -4)  # x + y - 4 = 0

intersect_point = l1.intersection(l2)

print("Intersection Point is:", intersect_point)
```

# How objects access atributes

# Reference variable
### Reference variable hold the onject
### We can create object without reference variable as well
### An object can have multiple reference variable
### Assiging a new reference variable to an existing object does not create a new object

```python
# object without a reference

class person:

  def __init__(self):
    self.name='Ghazala'
    self.population='1200000'
    self.capital='Karachi'

person()
```

```python
# multiple reference
p=person()
q=p
print(id(p))
print(id(q))
```

```python
# change atribute value with the help of 2D object
print(p.name)
print(q.name)
q.name = 'ghazal'
print(p.name)
print(q.name)
```
