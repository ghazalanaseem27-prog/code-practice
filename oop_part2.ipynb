# How objects access atributes

```python
import math

class point:
    def __init__(self, a, b):
        self.a_cor = a
        self.b_cor = b

    def __str__(self):
        return '<{},{}>'.format(self.a_cor, self.b_cor)

    def euclidean_distance(self, other):
        return ((self.a_cor - other.a_cor)**2 + (self.b_cor - other.b_cor)**2)**0.5

    def distance_from_origin(self):
        return (self.a_cor**2 + self.b_cor**2)**0.5

class line:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def __str__(self):
        return '{}x + {}y + {} = 0'.format(self.a, self.b, self.c)

    def point_on_line(line,point) :
      if line.a*point.a_cor+line.b*point.b_cor+line.c==0:
       return  'line on point'
      else:
        return 'not on line'
    def shortes_distance(line,point):
      return abs(line.a*point.a_cor+line.b*point.b_cor+line.c)/((line.a**2+line.b**2)**0.5)


p1 = point(3, 6)
p2 = point(4, 5)
l1 = line(1, 2, 3)

print(p1)
print(p1.euclidean_distance(p2))
print(p1.distance_from_origin())
print(l1)
print(l1.point_on_line(p1))
print(l1.shortes_distance(p1))
```

```python
class line:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

    def __str__(self):
        return '{}x + {}y + {} = 0'.format(self.a, self.b, self.c)

    # Naya function do lines ka intersection nikalne ke liye
    def intersection(self, other):
        # Denominator check (a1*b2 - a2*b1)
        denominator = (self.a * other.b) - (other.a * self.b)

        if denominator == 0:
            return "Lines are parallel, they do not intersect!"

        # Formula apply karna
        x = ((self.b * other.c) - (other.b * self.c)) / denominator
        y = ((self.c * other.a) - (other.c * self.a)) / denominator

        return point(x, y)

l1 = line(1, -1, 0)  # x - y = 0
l2 = line(1, 1, -4)  # x + y - 4 = 0

intersect_point = l1.intersection(l2)

print("Intersection Point is:", intersect_point)
```

# How to access atribute

```python
class Person:
    def __init__(self, name, population, capital):
        self.name = name
        self.population = population
        self.capital = capital

    def greet(self):
        if self.name == 'Ghazala':
            return f"Population is: {self.population}"
        else:
            return f"Capital is: {self.capital}"

p1 = Person('Ghazala', '1200000', 'Karachi')

print(p1.capital)
print(p1.greet())
print(p1.name)
print(p1.capital)
```
