```python
class Atm:
    def __init__(self):
        self.pin = ''
        # **1. PRIVATE ATTRIBUTE: Variable ke shuru mein __ lagane se yeh private ho gaya**
        self.__balance = 0
        self.menu()

    # these are untility methods->small work,they don't need object they can use class
    # **2. GETTER METHOD: Agar class ke baahir private balance DEKHNA ho**
    def get_balance(self):
        return self.__balance

    # **3. SETTER METHOD: Agar class ke baahir se private balance CHANGE karna ho**
    def set_balance(self, new_balance):
        # Hum yahan security check bhi laga sakte hain
        if type(new_balance) == int and new_balance >= 0:
            self.__balance = new_balance
            print("Balance updated successfully from outside!")
        else:
            print("Invalid amount!")

    def menu(self):
        user_input = input("""
        Hello, how would you like to proceed?
        1. Enter 1 to create pin
        2. Enter 2 for check balance
        3. Enter 3 to withdraw
        4. Enter 4 to deposit
        5. Enter 5 to Exit
        """)

        if user_input == '1':
            self.create_pin()
        elif user_input == '2':
            self.check_balance()
        elif user_input == '3':
            self.withdraw()
        elif user_input == '4':
            self.deposit()
        else:
            print('bye')

    def create_pin(self):
        self.pin = input('enter your pin: ')
        self.__balance = int(input('enter the balance: '))
        print('pin created successfully')
        self.menu()

    def check_balance(self):
        user_pin = input('enter your pin to check balance: ')
        if user_pin == self.pin:
            print('Your balance is:', self.__balance)
        else:
            print('invalid pin')
        self.menu()

    def withdraw(self):
        user_pin = input('enter your pin: ')
        if user_pin == self.pin:
            amount = int(input('enter amount to withdraw: '))
            if amount <= self.__balance:
                self.__balance -= amount
                print('withdraw successful')
            else:
                print('insufficient balance')
        else:
            print('invalid pin')
        self.menu()

    def deposit(self):
        user_pin = input('enter your pin: ')
        if user_pin == self.pin:
            amount = int(input('enter amount to deposit: '))
            self.__balance += amount
            print('deposit successful')
        else:
            print('invalid pin')
        self.menu()


# ========================================================
# HOW TO TEST AND CHANGE PRIVATE ATTRIBUTES FROM OUTSIDE
# ========================================================

# Object banaya (Aur menu ko cancel karne ke liye kuch bhi daba dein taaki baaki code chale)
obj = Atm()

# **Direct change karne ki koshish (Yeh kaam NAHI karega kyunki variable private hai):**
# obj.__balance = 5000000  <-- Isse error nahi aayega, par original balance change nahi hoga

# **Sahi tareeqa (Using Setter Method):**
obj.set_balance(5000) # Yeh private balance ko 5000 kar dega

# **Dekhne ka sahi tareeqa (Using Getter Method):**
print("Outside check - New Balance is:", obj.get_balance())
```

# Static variable (Vs instant variable)

```python
# static variable has same value for all objects
# instant variable has different value for all objects

# agr nam ka aga self laga ha to wo instance variable ha
# agr class ka name laga ho to wo static variable ha
```

```python
# need for static var
class Student:
    # **1. STATIC VARIABLE: Yeh class ke naam ke niche banta hai aur sab students ke liye SAME (Common) hota hai**
    college_name = "Government College"

    def __init__(self, name, roll_no):
        # **2. INSTANCE VARIABLES: Inke shuru mein 'self' hota hai aur yeh har student ke liye DIFFERENT (Alag) hote hain**
        self.name = name
        self.roll_no = roll_no

# --- Testing the variables ---

# Do alag objects (students) banaye
s1 = Student("Ghazala", 101)
s2 = Student("Ali", 102)

# **Instance Variables check karein (Dono ka data alag aayega):**
print(s1.name, s1.roll_no)  # Output: Ghazala 101
print(s2.name, s2.roll_no)  # Output: Ali 102

# **Static Variable check karein (Dono ke liye college ka naam bilkul SAME hai):**
print(s1.college_name)      # Output: Government College
print(s2.college_name)      # Output: Government College

# **Best Practice: Static variable ko hamesha Class ke naam se bhi call kiya ja sakta hai**
print(Student.college_name) # Output: Government College
```

```python
class Counter:
    # Static variable (Shared by all, same for all numbers)
    factor = 2

    def __init__(self, value):
        # Instance variable (Unique to each object)
        self.value = value

# Objects with different instance values
num1 = Counter(10)
num2 = Counter(50)

# Checking instance variables (Outputs will be different)
print(num1.value)  # Output: 10
print(num2.value)  # Output: 50

# Checking static variable (Outputs will be exactly the same)
print(num1.factor)     # Output: 2
print(Counter.factor)  # Output: 2
```
