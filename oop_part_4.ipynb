# Aggregation has a relationship

```python
# example
class Customer:
    def __init__(self, name, gender, address):
        self.name = name
        self.gender = gender
        self.address = address
    def print_address(self):
        print(self.address.get_city(), self.address.pin, self.address.state)

    def edit_profile(self,new_name,new_city,new_pin,new_state,):
       self.name = new_name
       self.address.edit_address(new_city,new_pin,new_state)

class Address:
    def __init__(self, city, pin, state):
        self.__city = city
        self.pin = pin
        self.state = state
  # we can,t access private attributes of aggregation for this we use getter
    def get_city(self):
        return self.__city

    def edit_address(self,new_city,new_pin,new_state):
      self.__city = new_city
      self.pin = new_pin
      self.state = new_state

# example object
add1 = Address('Gaggo',3456,'Vehari')
cust = Customer('Ghazala','Female',add1)
cust.print_address()
print(cust.gender)
print(cust.address.get_city())
print(cust.name)

cust.edit_profile('Gazal','Burarewala',3456,'Vehari')
cust.print_address()
print(cust.name)
```

# Super keyword
### super is a way of access to parent method

```python
# Method over riding
class phone:
  def __init__(self,brand,model,price):
    print('inside phone constructor')
    self.brand=brand
    self.model=model
    self.price=price
  def buy(self):
    print('buy phone')

class Smartphone(phone):
  def buy(self):
    print('buy smartphone')
    # syntax call to buy method parent
    super().buy()

s=Smartphone('Apple',4,400000)
s.buy()
```

```python
# super->construction

class Phone:
  def __init__(self,price,brand,camera):
    print('inside parent constructor')
    self.__price=price
    self.brand=brand
    self.camera=camera

class Smartphone(Phone):
  def __init__(self,price,brand,camera,os,ram):
    print('inside child constructor')
    super().__init__(price,brand,camera)
    self.os=os
    self.ram=ram

s=Smartphone(400000,'Apple',48,'Android',6)
print(s.os)
print(s.brand)
```

# Types of Inheritance

### * Single Inheritance
### * Multilevel Inheritance
### * Hierarchical Inheritance
### * Multiple Inheritance(Diamond Problem)
### * Hybrid Inheritance

### **Single**: 1 Parent -> 1 Child. (Child gets all properties of one parent).

### **Multilevel:** Grandparent -> Parent -> Child. (Like a family tree chain; child inherits from both parent and grandparent).

### **Hierarchical:** 1 Parent -> Multiple Children. (One main class shared by two or more completely different child classes).

### **Multiple:** Multiple Parents -> 1 Child. (A single child class inherits features from two or more independent parent classes at the same time).

### **Hybrid**: A mix of two or more inheritance types combined together (often creates a diamond structure).

```python
# single inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("Buying a phone")
class SmartPhone(Phone):
    pass

s = SmartPhone(20000, "Apple", "13mp")

s.buy()
```

```python
# multilevel inheritance
class Product:
    def review(self):
        print("Product review")

class Phone(Product):
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):
    pass

s = SmartPhone(20000, "Apple", "13mp")

s.buy()
s.review()
```

```python
# hierarchical inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("Buying a phone")

class SmartPhone(Phone):
    pass

class KeypadPhone(Phone):
    pass

s = SmartPhone(20000, "Apple", "13mp")
k = KeypadPhone(1000, "Nokia", "0.3mp")

s.buy()
k.buy()
```

```python
# multiple inheritance
class Phone:
    def __init__(self, price, brand, camera):
        print ("Inside phone constructor")
        self.__price = price
        self.brand = brand
        self.camera = camera

    def buy(self):
        print("Buying a phone")

class Product:
    def review(self):
        print("Customer review")

class SmartPhone(Phone, Product):
    pass

s = SmartPhone(20000, "Apple", "13mp")

s.buy()
s.review()
```

```python
# hybrid inheritance
class A:
    def min1(self):
        print("inside class A")

class B(A):
    def min2(self):
        print("inside class B")

class C(A):
    def min3(self):
        print("inside class C")

class D(B, C):
    def min4(self):
        print("inside class D")

d = D()
d.min1()
d.min2()
d.min3()
d.min4()
```

# Polymorphism Concepts
##  Polymorphism means having multiple forms. In Python, it is achieved through:

### 1. Method Overriding (which we covered)

### 2. Overloading (Method Overloading & Operator Overloading)

### 3. Polymorphism with Functions and Objects

```python
# method of overriding

class Animal:
    def speak(self):
        return "Animal makes a sound"

class Dog(Animal):
    # Same method name as parent
    def speak(self):
        return "Woof! Woof!"

# Objects banana
animal_obj = Animal()
dog_obj = Dog()

print(animal_obj.speak())  # Output: Animal makes a sound
print(dog_obj.speak())     # Output: Woof! Woof! (Overridden)
```

```python
# method overloading using default arguments
class Geometry:
    def area(self, a, b=0):
        if b == 0:
            print("Circle area:", 3.14 * (a ** 2))
        else:
            print("Rectangle area:", a * b)

obj = Geometry()
obj.area(4)     # acts as area of circle
obj.area(4, 5)  # acts as area of rectangle
```

```python
# operator overloading example

class Fraction:
    def __init__(self, num, den):
        self.num = num
        self.den = den

    # overloading the + operator
    def __add__(self, other):
        new_num = self.num * other.den + other.num * self.den
        new_den = self.den * other.den
        return f"{new_num}/{new_den}"

f1 = Fraction(1, 2)
f2 = Fraction(1, 3)
print(f1 + f2)  # output will be 5/6
```

```python
# Polymorphism with Functions and Objects

class Tomato:
    def type(self):
        print("Vegetable")
    def color(self):
        print("Red")

class Apple:
    def type(self):
        print("Fruit")
    def color(self):
        print("Red")

# Common function jo kisi bhi object ko accept kar sakta hai
def func(obj):
    obj.type()
    obj.color()

# Objects banana
obj_tomato = Tomato()
obj_apple = Apple()

# Ek hi function alag-alag objects ke sath alag behave kar raha hai
func(obj_tomato)
func(obj_apple)
```
