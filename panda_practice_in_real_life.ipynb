```python
import pandas as pd

# Create DataFrame
student = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain", "Usman"],
    "City": ["Lahore", "Karachi", "Lahore", "Karachi", "Lahore", "Karachi"],
    "Marks": [85, 92, 78, 90, 88, 80],
    "Age": [20, 21, 19, 22, 20, 21]
}

df = pd.DataFrame(student)

# Display DataFrame
print(df)

# Create groups by City
print(df.groupby("City"))

# Average Marks by City
print(df.groupby("City")["Marks"].mean())

# Total Marks by City
print(df.groupby("City")["Marks"].sum())

# Count students in each City
print(df.groupby("City")["Name"].count())

# Highest Marks in each City
print(df.groupby("City")["Marks"].max())

# Lowest Marks in each City
print(df.groupby("City")["Marks"].min())

# Multiple calculations together
print(df.groupby("City")["Marks"].agg(["sum", "mean", "max", "min", "count"]))

# Group by two columns
print(df.groupby(["City", "Age"])["Marks"].mean())

# Keep grouped column as normal column
print(df.groupby("City", as_index=False)["Marks"].mean())
```

```python
import pandas as pd

# Student information
df1 = pd.DataFrame({
    "ID": [1, 2, 3, 4],
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"]
})

# Student marks
df2 = pd.DataFrame({
    "ID": [1, 2, 3, 5],
    "Marks": [85, 92, 76, 90]
})

# Inner Merge (only matching IDs)
print(pd.merge(df1, df2, on="ID"))

# Left Join (keep all rows from df1)
print(pd.merge(df1, df2, on="ID", how="left"))

# Right Join (keep all rows from df2)
print(pd.merge(df1, df2, on="ID", how="right"))

# Outer Join (keep all rows from both DataFrames)
print(pd.merge(df1, df2, on="ID", how="outer"))

# Create two small DataFrames
df3 = pd.DataFrame({"Name": ["Ali", "Sara"]})
df4 = pd.DataFrame({"Name": ["Ahmed", "Ayesha"]})

# Combine rows
print(pd.concat([df3, df4]))

# Combine rows and reset index
print(pd.concat([df3, df4], ignore_index=True))

# Combine columns
name = pd.DataFrame({"Name": ["Ali", "Sara"]})
marks = pd.DataFrame({"Marks": [85, 92]})

print(pd.concat([name, marks], axis=1))
```

```python
import pandas as pd

# Create DataFrame
student = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Join_Date": [
        "2024-01-15",
        "2023-06-20",
        "2024-03-10",
        "2022-11-05",
        "2024-07-01"
    ]
}

df = pd.DataFrame(student)

# Display DataFrame
print(df)

# Check data types
print(df.dtypes)

# Convert string column to datetime
df["Join_Date"] = pd.to_datetime(df["Join_Date"])

# Check updated data types
print(df.dtypes)

# Extract Year
print(df["Join_Date"].dt.year)

# Extract Month
print(df["Join_Date"].dt.month)

# Extract Day
print(df["Join_Date"].dt.day)

# Extract weekday name
print(df["Join_Date"].dt.day_name())

# Extract month name
print(df["Join_Date"].dt.month_name())

# Create new Year, Month and Day columns
df["Year"] = df["Join_Date"].dt.year
df["Month"] = df["Join_Date"].dt.month
df["Day"] = df["Join_Date"].dt.day

print(df)

# Filter only 2024 records
print(df[df["Join_Date"].dt.year == 2024])

# Sort data by date
print(df.sort_values("Join_Date"))
```

```python
import pandas as pd

# Create DataFrame
sales = {
    "Salesperson": ["Ali", "Ali", "Sara", "Sara", "Ahmed", "Ahmed"],
    "City": ["Lahore", "Karachi", "Lahore", "Karachi", "Lahore", "Karachi"],
    "Product": ["Laptop", "Mobile", "Laptop", "Mobile", "Laptop", "Mobile"],
    "Sales": [50000, 30000, 55000, 35000, 45000, 25000]
}

df = pd.DataFrame(sales)

# Display DataFrame
print(df)

# Average Sales by City
print(df.pivot_table(values="Sales", index="City"))

# Total Sales by City
print(df.pivot_table(values="Sales", index="City", aggfunc="sum"))

# Maximum Sales by City
print(df.pivot_table(values="Sales", index="City", aggfunc="max"))

# Multiple calculations
print(df.pivot_table(
    values="Sales",
    index="City",
    aggfunc=["sum", "mean", "max", "min"]
))

# City and Product wise total sales
print(df.pivot_table(
    values="Sales",
    index=["City", "Product"],
    aggfunc="sum"
))

# Product becomes columns
print(df.pivot_table(
    values="Sales",
    index="City",
    columns="Product",
    aggfunc="sum"
))

# Replace NaN with 0
print(df.pivot_table(
    values="Sales",
    index="City",
    columns="Product",
    aggfunc="sum",
    fill_value=0
))

# Count Salesperson by City
print(pd.crosstab(df["City"], df["Salesperson"]))
```

```python
import pandas as pd

# Create DataFrame
student = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha", "Zain"],
    "Gender": ["M", "F", "M", "F", "M"],
    "Marks": [85, 92, 76, 90, 88]
}

df = pd.DataFrame(student)

# Display DataFrame
print(df)

# Convert M/F to Male/Female
df["Gender"] = df["Gender"].map({
    "M": "Male",
    "F": "Female"
})

print(df)

# Replace one value
print(df["Marks"].replace(76, 80))

# Replace multiple values
print(df["Marks"].replace({
    85: 86,
    92: 95
}))

# Add 5 bonus marks
print(df["Marks"].apply(lambda x: x + 5))

# Create Grade column
df["Grade"] = df["Marks"].apply(
    lambda x: "Pass" if x >= 80 else "Fail"
)

print(df)

# Create Total column (+10 marks)
df["Total"] = df["Marks"].apply(lambda x: x + 10)

print(df)
```
