# sequence sum
1/1! + 2/2! + 3/3! +.....

```python
n = int(input('enter n'))
result = 0
fact = 1
for i in range(1,n+1):
  fact = fact * i
  result = result + i/fact
print(result)

```

```python
raw = int(input("enter number of rows"))
for i in range(1,raw+1):
  for j in range(1,i+1):
    print(j,end='')
  for k in range(i-1,0,-1):
    print(k,end='')
  print()
```

# break statemet

```python

for i in range(1,10):
  if i == 5:
    break
  print(i)
```

```python
lower = int(input('enter lower range'))
upper = int(input('enter upper range'))
for i in range(lower,upper+1):
  for j in range(2,i):
    if i%j == 0:
      break
  else:
    print(i)
```

#continue statement
