#Accessing  substrings from string

```python
#  positive indexing

s= 'hellow world'
print(s[1])

# slicing index

print(s[0:5])
print(s[0:])
```

```python
# negative indexing

s = 'hellow world'
print(s[-1])

#slicing index

print(s[-5:-1])
```

```python
print(s[::-1])
```
