# if-else in python

```python
email = input("enter email")
password = input("enter password")
if email == "ghazala@gmail.com" and password == "1234":
  print("welcome")
else:
  print("incorrect")
```

# use if-else , elif and nested in one program

```python
email = input("enter email")
password = input("enter password")
if email == "Ghazala@gmail.com" and password == "1234":
  print("welcome")
elif email == "Ghazala@gmail.com" and password != "1234":
  print("try again")
  password = input("enter password again")
  if password == "1234":
    print("welcome")
  else:
    print("not correct")
else:
  print("correct")
```

# menu-drive *calculator*

```python
num1 = input("enter 1st num :")
snum = input("enter 2st num")
op = input("enter operator")
if op == "+":
    print(int(num1) + int(snum))
elif op == "-":
    print(int(num1) - int(snum))
elif op == "*":
  print(int(num1) * int(snum))
else :
  print(int(num1) / int(snum))
```

#generate a random integer between 1 and 100

```python
import random
jackpot =  random.randint(1,100)
guess = int(input("guess num"))
count = 1
while guess!=jackpot:
  if guess<jackpot:
    print("wrong! guess higher")
  else:
    print("wrong! guess lower")
  guess = int(input("guess"))
  count += 1
else:
  print("guess correct")
  print("attempts,count")
```
