```python
# Count the frequency of a particular character in a provided string.
# Eg 'hello how are you' is the string, the frequency of h in this string is 2

s = input('enter the string')
term = input('what would you like to search')

counter = 0
for i in s:
  if i == term:
    counter += 1

print('frequency', counter)
```

```python
# Write a program which can remove a particular character from a string.

s = input('enter the string')
term = input('what would you like to remove')

result = ''

for i in s:
  if i != term:
    result = result + i

print(result)
```

```python
# Write a program that can check whether a given string is palindrome or not

s = input('enter the string')
flag = True
for i in range(0, len(s)//2):
  if s[i] != s[len(s) - i - 1]:
    flag = False
    print('Not a palindrome')
    break

if flag == True:
  print('Palindrome')
```

```python
# Split the string without using the split() function

s = input('enter the string')

L = []
temp = ''
for i in s:
  if i == ' ':
    L.append(temp)
    temp = ''
  else:
    temp = temp + i

L.append(temp)
print(L)
```

```python
# Convert given input string to Title Case without using the title() function

s = input('enter the string')

L = []
for i in s.split():
  L.append(i[0].upper() + i[1:].lower())

print(" ".join(L))
```

```python
# Convert given integer to string without using the str() function


number = int(input('enter the number'))

digits = '0123456789'
result = ''
while number != 0:
  digit = number % 10
  result = digits[digit] + result
  number = number // 10

print(result)
print(type(result))
```
