```python
# arithmetic operators

print(10 + 20)
print(10 - 20)
print(10 * 20)
print(10 / 20)
print(10 // 20) # for integer devision
print(10 % 20)
print(10 ** 2) # for square
```

```python
# relational operators

print(10 > 20)
print(10 < 20)
print(10 >= 20)
print(10 <= 20)
print(10 == 20)
print(10 != 20)
```

```python
# (IMP)   logical operators

print(10 > 20 and 10 < 20)
print(10 > 20 or 10 < 20)
print(not(10 > 20))
```

```python
# bitwise operators

print(2 & 3) # bitsise and
print(2 | 3) # bitwise or
print(2 ^ 3) # bitwise xor
print(~3) # bitwise not
print(2 >> 3) # bitwisw left shif
print(2 << 3)# bitwise right shift
```

# Loops in python
### . Need for loop
### . While loop
### . For loop

Here is the difference between them in a short and simple way:

* **`while` loop:** Used when you **don't know** exactly how many times the loop will run. It keeps running as long as a specific condition is True (e.g., *keep accepting numbers until the user enters 0*).
* **`for` loop:** Used when you **know** exactly how many times the loop should run, or when you want to iterate through a sequence like a list, string, or a specific range of numbers.
* **Nested loop:** A loop **inside another loop** (a `for` or `while` loop within another `for` or `while` loop). The inner loop runs completely for every single turn of the outer loop (commonly used for working with grids, matrices, or patterns).

```python
# While loop example program -> to print table

num=int(input('enter num:'))
i=1
while i<=10:
    print(num*i)
    i+=1
```

```python
# program -> sum of all digits of all given numbers

num=input('enter num:')
total=10
i=1
while i<len(num):
    total+=int(num[i])
    i+=1
print(total)
```
