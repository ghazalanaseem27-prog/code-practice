# set operation

# Set function

```python
# len/sum/max/min/sorted
s={4,3,9,1,7,2,5}
len(s)
sum(s)
max(s)
min(s)
sorted(s)
```

```python
# union/update
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s1.update(s2)
print(s1)
print(s2)
s1.union(s2)
```

```python
# intersection/intersection-update
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s1.intersection(s2)
s1.intersection_update(s2)
print(s1)
print(s2)
```

```python
# difference/difference-update
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s1.difference(s2)
s1.difference_update(s2)
print(s1)
print(s2)
```

```python
# symmetric-difference/symmetric-difference-update
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s1.symmetric_difference(s2)
s1.symmetric_difference_update(s2)
print(s1)
print(s2)
```

```python
# isdisjoint/issubset/issuperset
s1={1,2,3,4,5}
s2={4,5,6,7,8}
s1.isdisjoint(s2)
s1.issubset(s2)
s1.issuperset(s2)
```

```python
# copy
s1={1,2,3,4,5}
s2=s1.copy()
print(s1)
print(s2)
```

# Dictionary comprehension

```python
# print 1 unm and their square
{i:i**2 for i in range(1,11)}
```

```python
# using existing dict
distances={'delhi':1000,'mumbai':2000,'goa':3000}
{key:value*0.05 for (key,value) in distances.items()}
```

```python
# using zip
days=['sun','mon','tue']
temp=[30,32,34]
{i:j for (i,j) in zip(days,temp)}
```

```python
from collections.abc import ValuesView
# using if condition
products={'a':200,'b':300,'c':400,'d':500}
{key:value for (key,value) in products.items() if value>0}
```

```python
# nested comprehension
# print table num from 2 to 4
{i:{j:i*j for j in range(1,11)}for i in range(2,5)}
```
