```python
color=set()

color.add('Red')
color.add('Black')
color.add('Blue')
color.add('Orange')
color.add('Green')
color.add('Black')

print(color)

color.remove('Orange')
print(color)

print(len(color))

```

```python
cricket_players={'Ahmad','Ali','Hamad','Asad','Ahad','Rafiq'}
football_players={'sarwar','Anwar','Masadaq','Ali','Ahad','Maqadas'}
print(cricket_players)
print(football_players)

print(cricket_players & football_players)

print( cricket_players | football_players)

print(cricket_players - football_players)

print(cricket_players ^ football_players)
```

```python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}
print(A & B)
print(A | B)
print(A - B)
print(A ^ B)
```

```python
# Dictionary examples
```

```python
car = {
    'brand': "Ford",
    'model': "Mustang",
    'year': 1964
}
print(car)

print(car.get('brand'))

print(car.get('color'))
```

```python
car = {
    'brand': "Ford",
    'model': "Mustang",
    'year': 1964
}
print(car)

print(car.keys())
print(len(car))
```

```python
car = {
    'brand': "Ford",
    'model': "Mustang",
    'year': 1964
}
print(car)

print(car.keys())
print(car.get('brand'))
print(car.values())
print(len(car))
print(car.items())

# this give none
print(car.update({'color': 'red'}))

# this update value
car.update({'year': 2020})
print(car)
```

```python
student = {
    "name": "Ali",
    "age": 20
}

student.update({"city": "Lahore"})
print(student)
student.update({"age": 22})
print(student)

```
