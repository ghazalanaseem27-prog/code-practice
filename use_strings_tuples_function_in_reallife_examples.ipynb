```python
subjects = ("English", "Math", "Science", "Computer", "Math")
print(subjects)

print(subjects[0])
print(subjects[-1])
print(subjects[0:4])
print(subjects.count("Math"))
print(subjects.index("Computer"))
```

```python
numbers = (10, 20, 30, 20, 40, 50, 20)
print(numbers)

print(numbers[1])
print(numbers[-1])
print(numbers[1:5])
print(numbers.count(20))
print(numbers.index(40))
print(len(numbers))
```
